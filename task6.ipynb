```python
graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D'],
    'C': ['A', 'D'],
    'D': ['B', 'C', 'E'],
    'E': ['D']
}

colors = {}

for vertex in graph:
    used_colors = set()

    for neighbor in graph[vertex]:
        if neighbor in colors:
            used_colors.add(colors[neighbor])

    color = 1

    while color in used_colors:
        color += 1

    colors[vertex] = color

print("Graph Coloring:")

for vertex in colors:
    print(vertex, "-> Color", colors[vertex])

print("\nNumber of colors used:", max(colors.values()))
```
