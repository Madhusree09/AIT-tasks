```python
def hill_climbing(values):
    current = 0

    while True:
        left = current - 1
        right = current + 1

        if left >= 0 and values[left] > values[current]:
            current = left


        elif right < len(values) and values[right] > values[current]:
            current = right

        else:
            break

    return values[current], current

values = [1, 3, 5, 7, 6, 4, 2]

best_value, position = hill_climbing(values)

print("Best value:", best_value)
print("Position:", position)
```
