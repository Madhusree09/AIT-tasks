```python
R1 = 30
R2 = 70

total = R1 + R2

p1 = R1 / total
p2 = R2 / total

percentage1 = p1 * 100
percentage2 = p2 * 100

print("Path 1 =", R1)
print("Path 2 =", R2)

print("Path 1 probability =", percentage1, "%")
print("Path 2 probability =", percentage2, "%")

if R1 < R2:
    print("Minimum path =", R1)
    print("Most ants follow the minimum path with",
          percentage1, "% chance")

elif R2 < R1:
    print("Minimum path =", R2)
    print("Most ants follow the minimum path with",
          percentage2, "% chance")

else:
    print("Both paths have equal probability")
```
